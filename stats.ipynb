```python
repos_d = defaultdict(int)
users_d = defaultdict(int)
with open("./data/mysql-2021-03-06/watchers.csv") as f:
    for line in tn(f):
        r, u, *o = line.split(",")
        repos_d[r] += 1
        users_d[u] += 1
len(repos_d), len(users_d)
```

```python
with open("./data/mysql-2021-03-06/followers.csv") as f:
    for line in tn(f):
        u1, u2, *o = line.split(",")
        users_d[u1] += 1
        users_d[u2] += 1
len(repos_d), len(users_d)
```
